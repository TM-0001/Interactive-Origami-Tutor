```python
# 青い三角形 ＋ 手が写っていない
import cv2
import numpy as np
import mediapipe as mp


# =========================================
# カメラ
# =========================================

cap = cv2.VideoCapture(0)


# =========================================
# MediaPipe Hands
# =========================================

mp_hands = mp.solutions.hands
mp_draw = mp.solutions.drawing_utils

hands = mp_hands.Hands(
    static_image_mode=False,
    max_num_hands=2,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)


while True:

    ret, frame = cap.read()

    if not ret:
        break


    # =========================================
    # ① 青色の検出
    # =========================================

    # BGR → HSV
    hsv = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2HSV
    )

    # 青色の範囲
    lower_blue = np.array([90, 50, 50])
    upper_blue = np.array([150, 255, 255])

    # 青色だけ抽出
    mask = cv2.inRange(
        hsv,
        lower_blue,
        upper_blue
    )


    # =========================================
    # ② マスクのノイズ除去
    # =========================================

    kernel = np.ones(
        (5, 5),
        np.uint8
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        kernel
    )

    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel
    )


    # =========================================
    # ③ 青色の輪郭を検出
    # =========================================

    contours, hierarchy = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    # =========================================
    # ④ 青い三角形か判定
    # =========================================

    is_triangle = False

    if len(contours) > 0:

        # 一番大きい青色の輪郭
        largest_contour = max(
            contours,
            key=cv2.contourArea
        )

        # 面積
        area = cv2.contourArea(
            largest_contour
        )

        # 輪郭の長さ
        perimeter = cv2.arcLength(
            largest_contour,
            True
        )

        # 多角形近似
        epsilon = 0.02 * perimeter

        approx = cv2.approxPolyDP(
            largest_contour,
            epsilon,
            True
        )

        # 3頂点なら三角形
        if len(approx) == 3:
            is_triangle = True


        # -----------------------------------------
        # 輪郭を描画
        # -----------------------------------------

        cv2.drawContours(
            frame,
            [largest_contour],
            -1,
            (0, 255, 0),
            3
        )


        # -----------------------------------------
        # バウンディングボックス
        # -----------------------------------------

        x, y, w, h = cv2.boundingRect(
            largest_contour
        )

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (255, 0, 0),
            2
        )


        # -----------------------------------------
        # 三角形なら赤色で表示
        # -----------------------------------------

        if is_triangle:

            cv2.drawContours(
                frame,
                [approx],
                -1,
                (0, 0, 255),
                5
            )


    # =========================================
    # ⑤ MediaPipeで手を検出
    # =========================================

    # BGR → RGB
    rgb = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    results = hands.process(rgb)


    # =========================================
    # ⑥ 手が写っているか判定
    # =========================================

    if results.multi_hand_landmarks:

        # 手が1つ以上写っている
        hands_visible = True

        # 手の骨格を描画
        for hand_landmarks in results.multi_hand_landmarks:

            mp_draw.draw_landmarks(
                frame,
                hand_landmarks,
                mp_hands.HAND_CONNECTIONS
            )

    else:

        # 手が写っていない
        hands_visible = False


    # =========================================
    # ⑦ 最終判定
    # =========================================

    # 青い三角形があり、
    # かつ手が写っていない
    if is_triangle and not hands_visible:

        result = True

    else:

        result = False


    # =========================================
    # ⑧ 結果を表示
    # =========================================

    print(result)


    # 画面上にも表示
    if result:

        cv2.putText(
            frame,
            "TRUE",
            (50, 80),
            cv2.FONT_HERSHEY_SIMPLEX,
            2,
            (0, 255, 0),
            4
        )

    else:

        cv2.putText(
            frame,
            "FALSE",
            (50, 80),
            cv2.FONT_HERSHEY_SIMPLEX,
            2,
            (0, 0, 255),
            4
        )


    # =========================================
    # ⑨ カメラ映像
    # =========================================

    cv2.imshow(
        "Camera",
        frame
    )


    # =========================================
    # ⑩ 青色マスク
    # =========================================

    cv2.imshow(
        "Mask",
        mask
    )


    # =========================================
    # qキーで終了
    # =========================================

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break


# =========================================
# 終了処理
# =========================================

cap.release()
hands.close()
cv2.destroyAllWindows()
```
